```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Demo dataset for the graphs : 
dataset  = sns.get_dataset_names()
dataset
```

```python
df = sns.load_dataset('diamonds')
```

```python
df.head()
```

```python
# Load Plotly : 
import plotly.express as px
```

```python
# Bar plot using plotly : 
px.bar(data_frame=df , x='cut')
```

```python
px.bar(df , x='cut' , color = 'color') # We can make the bar plot using the color method as well.
```

```python
px.bar(df , x = 'cut' , color='cut' , color_discrete_sequence=['red' , ' blue' , 'green' , 'black' ])
```

```python
data  = df.groupby(['cut']).size().rename('cut_count').reset_index()
```

```python
data.head()
```

```python
px.bar(data  , x='cut' , y='cut_count' , text='cut_count')
```
